```python
import json
import seaborn as sns
import numpy as np

import matplotlib.pyplot as plt

# Path to the jsonl file
file_path = "/root/workspace/LongSumm/jaehyeok/datasets/pairs_datasets.jsonl"

# Lists to store the ratios
ratios = []

# Read the jsonl file
with open(file_path, 'r') as file:
    for line in file:
        data = json.loads(line)
        section_sentences = data.get('section_sentences', [])
        best_index = data.get('best_index', 0)
        
        # Calculate ratio only if section_sentences is not empty
        if len(section_sentences) > 0:
            ratio = best_index / len(section_sentences)
            ratios.append(ratio)

# Plotting the distribution
plt.figure(figsize=(10, 6))
# Plot the histogram
ax = sns.histplot(ratios, kde=True, bins=30)
plt.title('Distribution of best_index/len(section_sentences)')
plt.xlabel('Ratio (best_index/len(section_sentences))')
plt.ylabel('Frequency')
plt.grid(True, alpha=0.3)

# Add mean and median lines
plt.axvline(np.mean(ratios), color='red', linestyle='--', label=f'Mean: {np.mean(ratios):.3f}')
plt.axvline(np.median(ratios), color='green', linestyle='--', label=f'Median: {np.median(ratios):.3f}')

# Create a secondary y-axis for proportions
ax2 = ax.twinx()
# Get the counts from the primary axis
counts = ax.patches
total = len(ratios)
# Set the secondary y-axis to show proportions
ax2.set_ylim(0, max([p.get_height() for p in counts]) / total)
ax2.set_ylabel('Proportion')
ax2.set_yticklabels([f'{x:.1%}' for x in ax2.get_yticks()])

plt.legend()
plt.tight_layout()
plt.show()

# Print some statistics
print(f"Number of entries: {len(ratios)}")
print(f"Mean ratio: {np.mean(ratios):.3f}")
print(f"Median ratio: {np.median(ratios):.3f}")
print(f"Min ratio: {min(ratios):.3f}")
print(f"Max ratio: {max(ratios):.3f}")
```

```python
# Lists to store the similarities
similarities = []

# Read the jsonl file
with open(file_path, 'r') as file:
    for line in file:
        data = json.loads(line)
        similarity = data.get('similarity', 0)
        similarities.append(similarity)

# Plotting the distribution
plt.figure(figsize=(10, 6))
# Plot the histogram
ax = sns.histplot(similarities, kde=True, bins=30)
plt.title('Distribution of Similarity Values')
plt.xlabel('Similarity')
plt.ylabel('Frequency')
plt.grid(True, alpha=0.3)

# Add mean and median lines
plt.axvline(np.mean(similarities), color='red', linestyle='--', label=f'Mean: {np.mean(similarities):.3f}')
plt.axvline(np.median(similarities), color='green', linestyle='--', label=f'Median: {np.median(similarities):.3f}')

# Create a secondary y-axis for proportions
ax2 = ax.twinx()
# Get the counts from the primary axis
counts = ax.patches
total = len(similarities)
# Set the secondary y-axis to show proportions
ax2.set_ylim(0, max([p.get_height() for p in counts]) / total)
ax2.set_ylabel('Proportion')
ax2.set_yticklabels([f'{x:.1%}' for x in ax2.get_yticks()])

plt.legend()
plt.tight_layout()
plt.show()

# Print some statistics
print(f"Number of entries: {len(similarities)}")
print(f"Mean similarity: {np.mean(similarities):.3f}")

print(f"Min similarity: {min(similarities):.3f}")
print(f"10th percentile: {np.percentile(similarities, 10):.3f}")
print(f"25th percentile: {np.percentile(similarities, 25):.3f}")
print(f"Median similarity: {np.median(similarities):.3f}")
print(f"75th percentile: {np.percentile(similarities, 75):.3f}")
print(f"90th percentile: {np.percentile(similarities, 90):.3f}")
print(f"Max similarity: {max(similarities):.3f}")
```
